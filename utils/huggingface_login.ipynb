# Hugging Face login

**Run this once, before any `benchmarking/` or `experiments/e1*/` notebook.** Every model here loads from Hugging Face via `from_pretrained(...)`, and several (Gemma especially) are gated -- they require your own account, your own access token, and that model's license accepted on its own page, none of which can be skipped.

Run the cell below and paste your token into the prompt it shows (the input is hidden, not echoed to the screen -- never type the token as literal code in a cell, since notebooks get saved and sometimes committed). It caches to `~/.cache/huggingface/token` on this machine, so this notebook only needs to run once per machine/account, not once per model or per session. Every other notebook checks for that cache automatically.

If a model notebook still fails after this with a gated-repo or license error, that means your account specifically hasn't accepted *that* model's license yet -- visit its page on huggingface.co and accept it there; logging in again here won't fix that part, it's a separate one-time step per model.

```python
from getpass import getpass
from huggingface_hub import login, whoami

try:
    print(f"Already logged in as {whoami()['name']}")
except Exception:
    login(token=getpass("Hugging Face token (input hidden): "))
    print(f"Logged in as {whoami()['name']}")
```
