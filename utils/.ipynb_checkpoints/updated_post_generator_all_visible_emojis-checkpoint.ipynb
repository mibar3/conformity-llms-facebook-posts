##  Individual creation of Facebook post  (option 1 - not overlapping reactions)
Includes:
- replacement of dislike by comment button
- likes and reactions (overlapping)
- nr shares and comments

```python
import os
os.makedirs(f"spotify_pie_plot/pie_plot_posts/baselines/incorrect/html", exist_ok=True)
```

```python
from utils.post_generator_all_visible_emojis import generate_facebook_post, display_facebook_post

PROFILE_NAME = "Andy Bara" # Andy Bara  or other gender neutral names
POST_TEXT = "The 2026 music genres distribution has just been released. Looks like Latin was more popular than Pop this year!"
POST_TIME = "Today at 2:43 PM"
VERIFIED = False  # Set to True to show verified badge

PROFILE_IMAGE_PATH =  None

# missing endearing emoji!
REACTION_DEFS = {
    "like":  ("👍", "#1877f2"),
    "love":  ("❤️", "#f33e58"),
    "haha":  ("😆", "#f7b928"),
    "wow":   ("😮", "#f7b928"),
    "sad":   ("😢", "#f7b928"),
    "angry": ("😡", "#e9710f"),
}

REACTIONS = {
    "like":  0,
    "love":  0,
    "haha":  0,
    "wow":   0,
    "sad":   0,
    "angry": 0,
}
COMMENT_COUNT = 0
SHARE_COUNT   = 0

for i in range(1, 2):
    POST_IMAGE_PATH = f"100_pie_charts/spotify_genre_pie_chart_{i:03d}.png"
    output_file = f"spotify_pie_plot/pie_plot_posts/baselines/incorrect/html/{i:03d}.html"
    
    html_content = generate_facebook_post(
        profile_name=PROFILE_NAME,
        post_text=POST_TEXT,
        post_time=POST_TIME,
        reactions=REACTIONS,
        comment_count=COMMENT_COUNT,
        share_count=SHARE_COUNT,
        post_image_path=POST_IMAGE_PATH,
        output_file=output_file,
        verified=VERIFIED,
    )
    display_facebook_post(html_content)
```
