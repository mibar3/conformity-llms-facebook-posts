# Benchmarking findings - Original plot

**Test 1 — test-1-gn-news-extensive (the extensive multi-question pilot)**
Asks 14 separate questions per image: profile name, profile verification, every individual engagement count (likes, comments, shares, and all 5 reaction types), the chart's two percentage values, chart colors, largest slice, and post_claim_correct (does the text match the chart). Run on only 4 images (1 gender-neutral profile + 1 "Dr." authority profile, × correct/incorrect), across 6 prompt wordings (v1–v6). Purpose: an extensive first pass to find the best-performing claim-verification wording, at the cost of a tiny sample.

**Test 2 — test-2-gn-claim-only (dedicated, zero-pressure claim check)**
Single question only: "does the post's text accurately describe the chart?" No engagement metrics shown — plain baseline images. 50 correct/50 incorrect pairs (100 images), gender-neutral profile only, same 6 wordings as Test 1, plus a two-call variant (v7–v9: extract the chart's values first, then verify the claim against them — the same two-call pattern later reused for the E1 two-step pilot). This is the true zero-social-pressure competence test.

**Test 3 — test-3-gn-percentages (perception check, not claim check)**
Reuses the exact same 50 image pairs as Test 2, but instead asks the model to read off the chart's two percentage values (Pop%, Latin%) directly — no claim judgment involved. Purpose, per the notebook's own rationale: direct comparability with Test 2 on identical images, so you can cross-reference per image ("read the percentages correctly but still got the claim wrong") — separating perception failure from reasoning failure.

**Test 4 — test-4-metrics-realistic-claim-only (claim check, with engagement now shown)**
Same single claim-verification question as Test 2, but now the post displays visible engagement metrics at 6 magnitude levels (10 → 1,000,000), with reactions distributed "realistically" (log-scaled across the 5 reaction types, approximating real engagement patterns) rather than split evenly. Purpose: does a single post's claim-verification accuracy shift once fake popularity is visible, isolated from the paired A/B framing used in the main E1 study. Worth noting for your methods section: an earlier "uniform reactions" version of this test (every reaction type set to the same count) was built and explicitly discarded as unrealistic before this log-scaled version replaced it (benchmarking/gemma4-12b-benchmarking.ipynb, cell 27, marked "old 4) ... will probably keep out because unrealistic").

## Test 1: extensive multi-question pilot

```python
from pathlib import Path
from utils.quanti_benchmarking_1_analysis import run_accuracy_analysis

BASE_DIR = Path().resolve()
# Updated to include "-news-" to match your folder structure
EXPERIMENT_NAME = "test-1-gn-news-extensive"

MODELS = [
    "gemma4-12b",
    "gemma-e4b",
    "ministral-3-14b",
    "ministral-3-8b",
    "qwen3-vl-4b",
    "qwen3-vl-8b"
]

for model_name in MODELS:
    print(f"\n" + "=" * 50)
    print(f" Running Accuracy Analysis for Model: {model_name} ")
    print("=" * 50)
    try:
        run_accuracy_analysis(BASE_DIR, EXPERIMENT_NAME, model_name)
    except Exception as e:
        print(f"Error processing {model_name}: {e}")
```

```python
import pandas as pd
import numpy as np

for model in MODELS:
    question_scores = {}
    
    # 1. Collect true/false scores across v1 to v6
    for v in range(1, 7):
        file_name = f"accuracy_scores_v{v}.csv"
        file_path = (
            BASE_DIR 
            / "outputs" 
            / model 
            / "quantitative" 
            / EXPERIMENT_NAME 
            / f"v{v}" 
            / file_name
        )
        
        if file_path.exists():
            df = pd.read_csv(file_path) 
            
            # Find all the columns your script generated for individual questions
            correct_cols = [c for c in df.columns if c.endswith("_correct") and c != "all_correct"]
            
            for col in correct_cols:
                question_name = col.replace("_correct", "")
                
                # Convert boolean True/False to 100.0 / 0.0
                scores = df[col].astype(float) * 100.0
                
                if question_name not in question_scores:
                    question_scores[question_name] = []
                # Append all image scores for this version
                question_scores[question_name].extend(scores.tolist())

    # 2. Filter questions strictly below 100% average
    struggling_questions = []
    for question, scores in question_scores.items():
        if scores:  # ensure we actually found data
            avg_acc = np.mean(scores)
            if avg_acc < 100.0:
                struggling_questions.append({
                    "Question": question,
                    "Avg Accuracy (%)": round(avg_acc, 2),
                    "Total Evaluations": len(scores) # Shows total number of images x versions evaluated
                })

    # 3. Print the overview
    print(f"\n{'='*60}\n STRUGGLING QUESTIONS: {model}\n{'='*60}")
    if struggling_questions:
        summary_df = pd.DataFrame(struggling_questions).sort_values(by="Avg Accuracy (%)")
        print(summary_df.to_string(index=False))
    else:
        print("No questions under 100% accuracy found (or files missing).")
```

```python
for model in MODELS:
    # find questions this model scored exactly 0.00% on
    question_scores = {}
    for v in range(1, 7):
        file_path = BASE_DIR / "outputs" / model / "quantitative" / EXPERIMENT_NAME / f"v{v}" / f"accuracy_scores_v{v}.csv"
        if not file_path.exists():
            continue
        df = pd.read_csv(file_path)
        correct_cols = [c for c in df.columns if c.endswith("_correct") and c != "all_correct"]
        for col in correct_cols:
            question_name = col.replace("_correct", "")
            scores = df[col].astype(float) * 100.0
            question_scores.setdefault(question_name, []).extend(scores.tolist())

    zero_questions = [q for q, scores in question_scores.items() if scores and sum(scores) == 0.0]

    print(f"\n{'='*70}\n RAW OUTPUTS FOR 0.00% CATEGORIES: {model}\n{'='*70}")
    if not zero_questions:
        print("  No categories scored exactly 0.00%.")
        continue

    for q in zero_questions:
        pred_col, true_col = f"{q}_pred", f"{q}_true"
        print(f"\n--- {q} ---")
        for v in range(1, 7):
            file_path = BASE_DIR / "outputs" / model / "quantitative" / EXPERIMENT_NAME / f"v{v}" / f"accuracy_scores_v{v}.csv"
            if not file_path.exists():
                continue
            df = pd.read_csv(file_path)
            if pred_col not in df.columns or true_col not in df.columns:
                continue
            for _, row in df.iterrows():
                print(f"  v{v} | {row['image']:22s} model said: {row[pred_col]!r:10}  ground truth: {row[true_col]!r}")
```

```python
import pandas as pd
import numpy as np

claim_accuracy_rows = []

for model in MODELS:
    variant_scores = {"correct": [], "incorrect": []}

    for v in range(1, 7):
        file_path = (
            BASE_DIR
            / "outputs"
            / model
            / "quantitative"
            / EXPERIMENT_NAME
            / f"v{v}"
            / f"accuracy_scores_v{v}.csv"
        )

        if not file_path.exists():
            continue

        df = pd.read_csv(file_path)
        if "post_claim_correct_correct" not in df.columns:
            continue

        for variant, group in df.groupby("variant"):
            scores = group["post_claim_correct_correct"].astype(float) * 100.0
            variant_scores.setdefault(variant, []).extend(scores.tolist())

    row = {"Model": model}
    for variant in ("correct", "incorrect"):
        scores = variant_scores.get(variant, [])
        row[f"{variant}_post_avg_accuracy_%"] = round(np.mean(scores), 2) if scores else np.nan
        row[f"{variant}_post_n"] = len(scores)
    if variant_scores.get("correct") and variant_scores.get("incorrect"):
        row["gap_incorrect_minus_correct_pts"] = round(
            row["incorrect_post_avg_accuracy_%"] - row["correct_post_avg_accuracy_%"], 2
        )
    claim_accuracy_rows.append(row)

claim_accuracy_df = pd.DataFrame(claim_accuracy_rows)
print(f"\n{'='*60}\n POST_CLAIM_CORRECT ACCURACY BY POST VARIANT\n{'='*60}")
print(claim_accuracy_df.to_string(index=False))
```

**gap_incorrect_minus_correct_pts** : Positive means the model is more accurate on incorrect posts than correct ones, across all six models in one table.

 # Test 2: Zero-pressure claim verification

```python
TEST2_NAME = "test-2-gn-claim-only"
TEST2_VERSIONS = range(1, 5)  # restricted to v1-v4 only, to match Test 4's wordings

test2_rows = []
for model in MODELS:
    variant_scores = {"correct": [], "incorrect": []}
    for v in TEST2_VERSIONS:
        file_path = BASE_DIR / "outputs" / model / "quantitative" / TEST2_NAME / f"v{v}" / f"accuracy_scores_v{v}.csv"
        if not file_path.exists():
            continue
        df = pd.read_csv(file_path)
        for variant, group in df.groupby("ground_truth"):
            variant_scores.setdefault(variant, []).extend((group["correct"].astype(bool) * 100.0).tolist())
    row = {"Model": model}
    for variant in ("correct", "incorrect"):
        scores = variant_scores.get(variant, [])
        row[f"{variant}_post_avg_accuracy_%"] = round(np.mean(scores), 2) if scores else np.nan
        row[f"{variant}_post_n"] = len(scores)
    all_scores = variant_scores["correct"] + variant_scores["incorrect"]
    row["pooled_avg_accuracy_%"] = round(np.mean(all_scores), 2) if all_scores else np.nan
    if variant_scores.get("correct") and variant_scores.get("incorrect"):
        row["gap_incorrect_minus_correct_pts"] = round(
            row["incorrect_post_avg_accuracy_%"] - row["correct_post_avg_accuracy_%"], 2
        )
    test2_rows.append(row)

test2_accuracy_df = pd.DataFrame(test2_rows)
print(f"\n{'='*70}\n TEST 2 — CLAIM VERIFICATION, NO ENGAGEMENT (v1-v4 only, matched to Test 4)\n{'='*70}")
print(test2_accuracy_df.to_string(index=False))
```

# Test 3: Chart percentage-reading accuracy

```python
TEST3_NAME = "test-3-gn-percentages"

test3_rows = []
for model in MODELS:
    f = BASE_DIR / "outputs" / model / "quantitative" / TEST3_NAME / f"accuracy_scores_{TEST3_NAME}.csv"
    if not f.exists():
        continue
    df = pd.read_csv(f)
    test3_rows.append({
        "Model": model,
        "pop_pct_accuracy_%": round(df["chart_percentages_pop_correct"].astype(bool).mean() * 100, 1),
        "latin_pct_accuracy_%": round(df["chart_percentages_latin_correct"].astype(bool).mean() * 100, 1),
        "both_correct_%": round((df["chart_percentages_pop_correct"].astype(bool) & df["chart_percentages_latin_correct"].astype(bool)).mean() * 100, 1),
        "n": len(df),
    })

test3_accuracy_df = pd.DataFrame(test3_rows)
print(f"\n{'='*60}\n TEST 3 — CHART PERCENTAGE-READING ACCURACY\n{'='*60}")
print(test3_accuracy_df.to_string(index=False))
```

# Test 4: Claim verification with engagement shown

Same single claim-verification question as Test 2, but the post now displays visible engagement
metrics at 6 magnitude levels (10 to 1,000,000), realistic (log-scaled) reaction distribution.
n=600 per wording version (v1-v4 only -- no v5/v6 for this test), 2,400 trials per model pooled.
Used as the comparison point against Test 2: does introducing fake popularity change
claim-verification accuracy on the same underlying task?

```python
TEST4_NAME = "test-4-metrics-realistic-claim-only"
TEST4_VERSIONS = range(1, 5)  # only v1-v4 exist for this test
SCALE_VALUES = [10, 100, 1000, 10000, 100000, 1000000]

test4_rows = []
test4_scale_rows = []
for model in MODELS:
    variant_scores = {"correct": [], "incorrect": []}
    scale_scores = {s: [] for s in SCALE_VALUES}
    for v in TEST4_VERSIONS:
        file_path = BASE_DIR / "outputs" / model / "quantitative" / TEST4_NAME / f"v{v}" / f"accuracy_scores_v{v}.csv"
        if not file_path.exists():
            continue
        df = pd.read_csv(file_path)
        for variant, group in df.groupby("ground_truth"):
            variant_scores.setdefault(variant, []).extend((group["correct"].astype(bool) * 100.0).tolist())
        for scale, group in df.groupby("scale_value"):
            if int(scale) in scale_scores:
                scale_scores[int(scale)].extend((group["correct"].astype(bool) * 100.0).tolist())

    row = {"Model": model}
    for variant in ("correct", "incorrect"):
        scores = variant_scores.get(variant, [])
        row[f"{variant}_post_avg_accuracy_%"] = round(np.mean(scores), 2) if scores else np.nan
        row[f"{variant}_post_n"] = len(scores)
    all_scores = variant_scores["correct"] + variant_scores["incorrect"]
    row["pooled_avg_accuracy_%"] = round(np.mean(all_scores), 2) if all_scores else np.nan
    if variant_scores.get("correct") and variant_scores.get("incorrect"):
        row["gap_incorrect_minus_correct_pts"] = round(
            row["incorrect_post_avg_accuracy_%"] - row["correct_post_avg_accuracy_%"], 2
        )
    test4_rows.append(row)

    scale_row = {"Model": model}
    for s in SCALE_VALUES:
        scores = scale_scores[s]
        scale_row[f"scale_{s}_%"] = round(np.mean(scores), 2) if scores else np.nan
    test4_scale_rows.append(scale_row)

test4_accuracy_df = pd.DataFrame(test4_rows)
test4_scale_df = pd.DataFrame(test4_scale_rows)

print(f"\n{'='*70}\n TEST 4 — CLAIM VERIFICATION WITH ENGAGEMENT SHOWN (pooled across all scales)\n{'='*70}")
print(test4_accuracy_df.to_string(index=False))

print(f"\n{'='*80}\n TEST 4 — ACCURACY BY ENGAGEMENT SCALE (does magnitude matter within Test 4?)\n{'='*80}")
print(test4_scale_df.to_string(index=False))
```

# Test 2 vs. Test 4: does introducing engagement change claim-verification accuracy?
