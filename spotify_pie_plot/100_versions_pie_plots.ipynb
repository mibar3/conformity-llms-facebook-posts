# Creating 100 different versions of the pie plot in the generated facebook post for randomization
Keeping user and text same

```python
import matplotlib.pyplot as plt
import numpy as np
import random
import os

# Base labels
labels = ["Pop", "Hip-Hop / Rap", "Rock", "Latin", "Electronic / Dance",
          "R&B / Soul", "Indie / Alternative", "Classical", "Other"]

# Color palettes to randomly pick from
color_palettes = [
    ["#1DB954", "#191414", "#e74c3c", "#f39c12", "#3498db", "#9b59b6", "#1abc9c", "#bdc3c7", "#e0e0e0"],
    ["#FF6B6B", "#4ECDC4", "#45B7D1", "#96CEB4", "#FFEAA7", "#DDA0DD", "#98D8C8", "#F7DC6F", "#BB8FCE"],
    ["#2C3E50", "#E74C3C", "#ECF0F1", "#3498DB", "#2ECC71", "#F39C12", "#9B59B6", "#1ABC9C", "#E67E22"],
    ["#FF9A9E", "#FAD0C4", "#A18CD1", "#FBC2EB", "#84FAB0", "#8FD3F4", "#D4FC79", "#96E6A1", "#FFDDE1"],
    ["#667EEA", "#764BA2", "#F093FB", "#F5576C", "#4FACFE", "#00F2FE", "#43E97B", "#38F9D7", "#FA709A"],
]

# Dark colors that need white text
dark_colors = {
    "#191414", "#000000", "#2c3e50", "#000", "#667eea", "#764ba2",
    "#191414", "#3498db", "#9b59b6", "#4facfe", "#f5576c"
}
```

# Pop 23.5% - Latin 11%

```python
# Fixed values
fixed_pop = 23.5
fixed_latin = 11.0

import os

generated_count = skipped_count = 0
for i in range(100):  # change to 100 for full run
    output_path = f"100_pie_charts/spotify_genre_pie_chart_{i+1:03d}.png"

    # If the chart already exists, skip regenerating it -- saves the randomization
    # and plotting work too, not just the save call.
    if os.path.exists(output_path):
        skipped_count += 1
        continue

    random.seed(i)
    np.random.seed(i)
    # Randomize non-fixed sizes ensuring no slice is under 3%
    remaining = 100 - fixed_pop - fixed_latin
    min_val = 3.0

    while True:
        rand_sizes = np.random.dirichlet(np.ones(7)) * remaining #generates values that always sum to exactly 1, which when multiplied by remaining (65.5) always sum to exactly 65.5
        # enforce minimum
        for j in range(len(rand_sizes)):
            if rand_sizes[j] < min_val:
                diff = min_val - rand_sizes[j]
                rand_sizes[j] = min_val
                rand_sizes[np.argmax(rand_sizes)] -= diff
        if all(s >= min_val for s in rand_sizes):
            break

    rand_sizes = np.round(rand_sizes, 1)
    # fix rounding drift
    rand_sizes[-1] += round(remaining - rand_sizes.sum(), 1)

    sizes = [fixed_pop, rand_sizes[0], rand_sizes[1], fixed_latin,
             rand_sizes[2], rand_sizes[3], rand_sizes[4], rand_sizes[5], rand_sizes[6]]

    # Random choices
    colors = random.choice(color_palettes)[:]
    random.shuffle(colors)

    startangle = random.randint(0, 360)

    # Random explode
    explode = [0.0] * 9
    explode[random.randint(0, 8)] = random.choice([0.0, 0.05, 0.1])
    explode = tuple(explode)

    pctdistance = random.uniform(0.70, 0.88)
    figsize = random.choice([(6, 6), (7, 7), (6.5, 6.5)])

    fig, ax = plt.subplots(figsize=figsize)
    wedges, texts, autotexts = ax.pie(
        sizes,
        labels=labels,
        colors=colors,
        explode=explode,
        autopct="%1.1f%%",
        startangle=startangle,
        pctdistance=pctdistance,
        wedgeprops=dict(edgecolor="white", linewidth=random.uniform(1.0, 2.5))
    )

    # Single font size for all percentage labels
    fontsize = random.randint(8, 11)
    for k, text in enumerate(autotexts):
        text.set_fontsize(fontsize)
        text.set_fontweight("bold")
        # white text on dark wedges
        if colors[k].lower() in dark_colors:
            text.set_color("white")
        else:
            text.set_color("black")

    ax.set_title("Music Genre Preferences of Spotify Users",
                 fontsize=random.randint(11, 14),
                 fontweight="bold", pad=20)

    plt.tight_layout()
    plt.savefig(output_path, dpi=150, bbox_inches="tight")
    plt.close()
    generated_count += 1
    print(f"Saved chart {i+1}")

print(f"Done! generated {generated_count}, skipped {skipped_count} (already existed)")
```
