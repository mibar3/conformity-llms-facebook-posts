This check confirms that " like", " scroll", " yes", and " no" each tokenize as a single token for every model. 

This is a requirement for the log-probability scoring method, which reads each candidate's probability directly off the first generated token's distribution, so a multi-token candidate would silently produce an incorrect score.

```python
from transformers import AutoProcessor

MODELS = [
    ("Gemma-12B",              "google/gemma-4-12B-it",                         {"padding_side": "left"}),
    ("Gemma-E4B",              "google/gemma-4-E4B-it",                         {"padding_side": "left"}),
    ("Qwen3-VL-4B",            "Qwen/Qwen3-VL-4B-Instruct",                     {}),
    ("Qwen3-VL-8B",            "Qwen/Qwen3-VL-8B-Instruct",                     {}),
    #("Pixtral-12B",            "mistral-community/pixtral-12b",                 {}),
    #("Mistral-Small-3.1-24B",  "mistralai/Mistral-Small-3.1-24B-Instruct-2503", {"fix_mistral_regex": True}),
    ("Ministral-3-8B",         "mistralai/Ministral-3-8B-Instruct-2512-BF16",   {"fix_mistral_regex": True}),
    ("Ministral-3-14B",        "mistralai/Ministral-3-14B-Instruct-2512-BF16",  {"fix_mistral_regex": True}),
]

for label, model_id, kwargs in MODELS:
    tok = AutoProcessor.from_pretrained(model_id, **kwargs).tokenizer
    for word in [" like", " scroll", " yes", " no"]:
        ids = tok(word, add_special_tokens=False).input_ids
        flag = "OK" if len(ids) == 1 else "MULTI-TOKEN!"
        print(f"{label:25s} {word!r:10s} -> {ids} {flag}")
    print()
```
