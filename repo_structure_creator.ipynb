# Creating copy of repo structure

```python
import os
from pathlib import Path

# Explicitly set the root path to your repo
from pathlib import Path as _P
REPO_ROOT = str(_P().resolve())  # run from the repo root
output_file = f"{REPO_ROOT}/project_structure.txt"

# Directories to completely skip
ignore_dirs = {'.git', '__pycache__', 'venv', '.venv', 'outputs', 'node_modules', '.ipynb_checkpoints'}

# Target extensions to trigger the summary threshold
target_exts = {'.json', '.html', '.png'}

with open(output_file, 'w') as f:
    for root, dirs, files in os.walk(REPO_ROOT):
        # Modify dirs in-place to skip ignored directories
        dirs[:] = [d for d in dirs if d not in ignore_dirs]
        
        # Calculate indentation based on relative path
        rel_path = os.path.relpath(root, REPO_ROOT)
        if rel_path == '.':
            level = 0
            folder_name = "conformity-llms-facebook-posts/"
        else:
            level = rel_path.count(os.sep) + 1
            folder_name = os.path.basename(root) + "/"
            
        indent = '    ' * level
        f.write(f'{indent}{folder_name}\n')
        
        # Filter and sort visible files (skip hidden ones)
        visible_files = [file for file in sorted(files) if not file.startswith('.')]
        
        # Count how many relevant files (.json, .html, .png) are in this folder
        matching_count = sum(1 for file in visible_files if Path(file).suffix.lower() in target_exts)
        
        # Write the files inside
        subindent = '    ' * (level + 1)
        current_dir_name = os.path.basename(root)
        
        # Condition to summarize: >10 target files OR the folder name is purely numeric
        should_summarize = (matching_count > 10) or current_dir_name.isdigit()
        
        if should_summarize and len(visible_files) > 1:
            # Print the very first file as an example
            f.write(f'{subindent}{visible_files[0]}\n')
            # Print the summary line for the rest
            f.write(f'{subindent}... and {len(visible_files) - 1} more file(s) with similar structure\n')
        else:
            # Print all files normally
            for file in visible_files:
                f.write(f'{subindent}{file}\n')
            
print(f"✅ project_structure.txt has been created at: {output_file}")
```
